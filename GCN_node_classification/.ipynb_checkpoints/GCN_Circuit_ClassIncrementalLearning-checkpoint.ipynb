## Node Classification

#### One shot static learning demonstration using GCN

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import dgl
import dgl.nn as dglnn
import ast
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# -------------------------
# Load and Prepare Dataset
# -------------------------
df = pd.read_csv("all_circuits_features.csv")

print("\n?? Dataset Overview:")
print(df.info())

# Convert 'gate_type' to numerical labels
label_encoder = LabelEncoder()
df["gate_label"] = label_encoder.fit_transform(df["gate_type"])

# Feature columns
feature_columns = [
    "fan_in", "fan_out", "dist_to_output", "is_primary_input", "is_primary_output",
    "is_internal", "is_key_gate", "degree_centrality", "betweenness_centrality",
    "closeness_centrality", "clustering_coefficient", "avg_fan_in_neighbors", "avg_fan_out_neighbors"
]

# Drop NaN values and convert features to float
df = df.dropna(subset=feature_columns)
df[feature_columns] = df[feature_columns].astype(float)

# Normalize numerical features
scaler = StandardScaler()
df[feature_columns] = scaler.fit_transform(df[feature_columns])

# -------------------------
# Step 1: Extract Nodes
# -------------------------
nodes = df["node"].tolist()
node_to_id = {node: i for i, node in enumerate(nodes)}

# -------------------------
# Step 2: Extract Edges from fan-in relationships
# -------------------------
edges = []
for idx, row in df.iterrows():
    node_id = node_to_id[row["node"]]
    
    # Find nodes with fan-out > 0 (likely driving this gate)
    potential_sources = df[df["fan_out"] > 0]["node"].tolist()
    
    # Assign 'fan_in' number of edges
    sources = potential_sources[:int(row["fan_in"])] if int(row["fan_in"]) > 0 else []
    for src in sources:
        if src in node_to_id:
            edges.append((node_to_id[src], node_id))  # Directed edge from src ? node

# -------------------------
# Step 3: Validate Edge List
# -------------------------
print("\n?? Extracted", len(edges), "edges.")
if len(edges) == 0:
    print("\n? ERROR: No edges found! Check `fan_in` values.")
    print("\n?? First 10 rows of dataset:")
    print(df[["node", "fan_in", "fan_out"]].head(10))
    raise ValueError("Error: Edge list is empty. Please check `fan_in` and `fan_out` values.")

src_nodes, dst_nodes = zip(*edges) if edges else ([], [])
src_tensor = torch.tensor(src_nodes, dtype=torch.int64)
dst_tensor = torch.tensor(dst_nodes, dtype=torch.int64)

# Remove invalid edges (if any)
valid_edges = (src_tensor >= 0) & (dst_tensor >= 0) & (src_tensor < len(nodes)) & (dst_tensor < len(nodes))
src_tensor = src_tensor[valid_edges]
dst_tensor = dst_tensor[valid_edges]

# -------------------------
# Step 4: Create DGL Graph
# -------------------------
graph = dgl.graph((src_tensor, dst_tensor), num_nodes=len(nodes))
graph = dgl.add_self_loop(graph)  # Add self-loops to avoid zero in-degree

# Assign node features and labels
graph.ndata['features'] = torch.tensor(df[feature_columns].values, dtype=torch.float32)
graph.ndata['labels'] = torch.tensor(df["gate_label"].values, dtype=torch.long)

# -------------------------
# Train-Test Split
# -------------------------
train_mask, test_mask = train_test_split(range(len(nodes)), test_size=0.2, random_state=42)
train_mask = torch.tensor(train_mask, dtype=torch.long)
test_mask = torch.tensor(test_mask, dtype=torch.long)

# -------------------------
# Step 5: Define GCNN Model
# -------------------------
class GCN(nn.Module):
    def __init__(self, in_feats, hidden_feats, out_feats):
        super(GCN, self).__init__()
        self.conv1 = dglnn.GraphConv(in_feats, hidden_feats, allow_zero_in_degree=True)
        self.conv2 = dglnn.GraphConv(hidden_feats, out_feats, allow_zero_in_degree=True)

    def forward(self, g, inputs):
        x = self.conv1(g, inputs)
        x = torch.relu(x)
        x = self.conv2(g, x)
        return x

in_feats = len(feature_columns)
hidden_feats = 32
out_feats = len(label_encoder.classes_)
model = GCN(in_feats, hidden_feats, out_feats)
optimizer = optim.Adam(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

# -------------------------
# Step 6: Training Loop
# -------------------------
epochs = 50
for epoch in range(epochs):
    model.train()
    logits = model(graph, graph.ndata['features'])
    loss = loss_fn(logits[train_mask], graph.ndata['labels'][train_mask])

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if epoch % 10 == 0:
        print(f"Epoch {epoch}/{epochs}, Loss: {loss.item():.4f}")

# -------------------------
# Step 7: Evaluate Model
# -------------------------
model.eval()
with torch.no_grad():
    test_logits = model(graph, graph.ndata['features'])
    test_predictions = test_logits[test_mask].argmax(dim=1)
    test_accuracy = (test_predictions == graph.ndata['labels'][test_mask]).float().mean().item()
    print(f"\n? Test Accuracy: {test_accuracy * 100:.2f}%")
    
    train_predictions = test_logits[train_mask].argmax(dim=1)
    train_accuracy = (train_predictions == graph.ndata['labels'][train_mask]).float().mean().item()
    print(f"\n? Train Accuracy: {train_accuracy * 100:.2f}%")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# True and predicted labels
y_true = graph.ndata['labels'][test_mask].cpu().numpy()
y_pred = test_logits[test_mask].argmax(dim=1).cpu().numpy()

# Get back the original class names from your LabelEncoder
class_names = label_encoder.classes_

# Classification report with class names
print("\n=== Classification Report ===")
print(classification_report(y_true, y_pred, target_names=class_names, digits=4))

# Confusion matrix with class names
cm = confusion_matrix(y_true, y_pred)
print("\n=== Confusion Matrix ===")
print(pd.DataFrame(cm, index=class_names, columns=class_names))

# Optional: visualize as heatmap
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()
```

#### Static Neural Architecture

Static architecture: a GCN with 8 outputs (for the 8 gate types) from the start.

Incremental batches:
- and, not
- nor
- nand
- input
- output, or
- xor

Mechanisms:
- Naive fine-tuning
- Rehearsal buffer
- EWC
- Synaptic Intelligence (SI)
- Generative replay (stubbed as rehearsal)

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import dgl
import dgl.nn as dglnn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_fscore_support
from tabulate import tabulate

# -------------------------
# Load and Prepare Dataset
# -------------------------
df = pd.read_csv("all_circuits_features.csv")

# Label encoding
label_encoder = LabelEncoder()
df["gate_label"] = label_encoder.fit_transform(df["gate_type"])

# Feature columns
feature_columns = [
    "fan_in", "fan_out", "dist_to_output", "is_primary_input", "is_primary_output",
    "is_internal", "is_key_gate", "degree_centrality", "betweenness_centrality",
    "closeness_centrality", "clustering_coefficient", "avg_fan_in_neighbors", "avg_fan_out_neighbors"
]

# Clean and normalize
df = df.dropna(subset=feature_columns)
df[feature_columns] = df[feature_columns].astype(float)
scaler = StandardScaler()
df[feature_columns] = scaler.fit_transform(df[feature_columns])

# -------------------------
# Build graph from CSV (simple fan-in heuristic)
# -------------------------
nodes = df["node"].tolist()
node_to_id = {node: i for i, node in enumerate(nodes)}

edges = []
for _, row in df.iterrows():
    node_id = node_to_id[row["node"]]
    potential_sources = df[df["fan_out"] > 0]["node"].tolist()
    k = int(row["fan_in"]) if not np.isnan(row["fan_in"]) else 0
    sources = potential_sources[:max(k, 0)]
    for src in sources:
        if src in node_to_id:
            edges.append((node_to_id[src], node_id))

src_nodes, dst_nodes = zip(*edges) if edges else ([], [])
src_tensor = torch.tensor(src_nodes, dtype=torch.int64)
dst_tensor = torch.tensor(dst_nodes, dtype=torch.int64)
valid_edges = (src_tensor >= 0) & (dst_tensor >= 0) & (src_tensor < len(nodes)) & (dst_tensor < len(nodes))
src_tensor = src_tensor[valid_edges]
dst_tensor = dst_tensor[valid_edges]

graph = dgl.graph((src_tensor, dst_tensor), num_nodes=len(nodes))
graph = dgl.add_self_loop(graph)
graph.ndata['features'] = torch.tensor(df[feature_columns].values, dtype=torch.float32)
graph.ndata['labels'] = torch.tensor(df["gate_label"].values, dtype=torch.long)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
graph = graph.to(device)
graph.ndata['features'] = graph.ndata['features'].to(device)
graph.ndata['labels'] = graph.ndata['labels'].to(device)

# -------------------------
# Class-incremental batches
# -------------------------
class_to_label = {c: i for i, c in enumerate(label_encoder.classes_)}
class_batches = [
    ["and", "not"],   # Batch 1
    ["nor"],          # Batch 2
    ["nand"],         # Batch 3
    ["input"],        # Batch 4
    ["output", "or"], # Batch 5
    ["xor"]           # Batch 6
]

all_labels_np = graph.ndata['labels'].cpu().numpy()

def indices_for_classes(class_names):
    class_ids = [class_to_label[c] for c in class_names]
    mask = np.isin(all_labels_np, class_ids)
    return np.where(mask)[0]

batch_indices = [indices_for_classes(b) for b in class_batches]

# -------------------------
# GCN model (static 8-way)
# -------------------------
class GCN(nn.Module):
    def __init__(self, in_feats, hidden_feats, out_feats):
        super(GCN, self).__init__()
        self.conv1 = dglnn.GraphConv(in_feats, hidden_feats, allow_zero_in_degree=True)
        self.conv2 = dglnn.GraphConv(hidden_feats, out_feats, allow_zero_in_degree=True)
    def forward(self, g, inputs):
        x = self.conv1(g, inputs)
        x = torch.relu(x)
        x = self.conv2(g, x)
        return x

in_feats = graph.ndata['features'].shape[1]
hidden_feats = 64
out_feats = len(label_encoder.classes_)

# -------------------------
# Mechanisms
# -------------------------
def _to_list(idx):
    if isinstance(idx, np.ndarray): return idx.tolist()
    if isinstance(idx, list): return idx
    return list(idx)

class RehearsalBuffer:
    def __init__(self, size=2000, num_nodes=None):
        self.size = size
        self.data_idx = []
        self.num_nodes = num_nodes
    def add(self, idx):
        idx_list = _to_list(idx)
        self.data_idx.extend(idx_list)
        if self.num_nodes is not None:
            self.data_idx = [i for i in self.data_idx if 0 <= i < self.num_nodes]
        self.data_idx = list(dict.fromkeys(self.data_idx))
        if len(self.data_idx) > self.size:
            self.data_idx = np.random.choice(self.data_idx, self.size, replace=False).tolist()
    def get(self):
        return self.data_idx

class EWC:
    def __init__(self, model, lam=200.0):
        self.model = model
        self.lam = lam
        self.star_vars = None
        self.fisher = None
    def compute_fisher(self, g, features, labels, idx, sample_size=2000):
        idx_list = _to_list(idx)
        if len(idx_list) == 0: return
        sample_idx = np.random.choice(idx_list, min(sample_size, len(idx_list)), replace=False)
        sample_idx_t = torch.tensor(sample_idx, dtype=torch.long, device=device)
        self.model.train()
        logits = self.model(g, features)
        loss = nn.CrossEntropyLoss()(logits[sample_idx_t], labels[sample_idx_t])
        self.model.zero_grad()
        loss.backward()
        self.fisher = [p.grad.detach().cpu().numpy()**2 if p.grad is not None else np.zeros_like(p.detach().cpu().numpy())
                       for p in self.model.parameters()]
        self.star_vars = [p.detach().cpu().numpy().copy() for p in self.model.parameters()]
    def penalty(self, model):
        if self.fisher is None:
            return torch.tensor(0.0, device=device)
        total = 0.0
        for p, p_star, f in zip(model.parameters(), self.star_vars, self.fisher):
            total += np.sum(f * (p.detach().cpu().numpy() - p_star)**2)
        return torch.tensor((self.lam/2.0)*total, dtype=torch.float32, device=device)

class SITracker:
    def __init__(self, model, c=2.0, eps=1e-3):
        self.model = model
        self.c = c
        self.eps = eps
        self.prev_vars = [p.detach().cpu().numpy().copy() for p in model.parameters()]
        self.w_importance = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
        self.path_int = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
    def update(self, grads):
        for i,(g,p) in enumerate(zip(grads,self.model.parameters())):
            if g is not None:
                delta = p.detach().cpu().numpy() - self.prev_vars[i]
                self.path_int[i] += (-g.detach().cpu().numpy()) * delta
            self.prev_vars[i] = p.detach().cpu().numpy().copy()
    def consolidate(self):
        for i,p in enumerate(self.model.parameters()):
            delta = p.detach().cpu().numpy() - self.prev_vars[i]
            self.w_importance[i] += self.path_int[i]/(delta**2+self.eps)
            self.path_int[i] = np.zeros_like(p.detach().cpu().numpy())
    def penalty(self, model):
        total=0.0
        for i,p in enumerate(model.parameters()):
            total += np.sum(self.w_importance[i]*(p.detach().cpu().numpy()-self.prev_vars[i])**2)
        return torch.tensor(self.c*total, dtype=torch.float32, device=device)

# -------------------------
# Metrics helpers
# -------------------------
def compute_metrics(y_true, y_pred, class_names):
    labels = list(range(len(class_names)))
    acc = accuracy_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    pr, rc, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, average="macro", zero_division=0
    )
    return acc, cm, pr, rc, f1

def print_batch_report(mech, batch_num, batch_classes, acc, cm, pr, rc, f1, class_names):
    print(f"\n=== Mechanism: {mech} | After batch {batch_num} ({batch_classes}) ===")
    print(f"Accuracy: {acc:.4f}, Precision (macro): {pr:.4f}, Recall: {rc:.4f}, F1 (macro): {f1:.4f}")
    print(pd.DataFrame(cm, index=class_names, columns=class_names))

# -------------------------
# Training loop per mechanism
# -------------------------
def run_experiment(mech="naive", epochs_per_batch=5, lr=1e-3):
    model = GCN(in_feats, hidden_feats, out_feats).to(device)
    opt = optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.CrossEntropyLoss()

    buffer = RehearsalBuffer(2000, num_nodes=graph.num_nodes()) if mech in ["rehearsal","gen_replay"] else None
    ewc = EWC(model, lam=200.0) if mech=="ewc" else None
    si = SITracker(model, c=2.0) if mech=="si" else None

    results = {"acc":[], "prec":[], "rec":[], "f1":[], "cm":[]}

    features = graph.ndata['features']
    labels = graph.ndata['labels']

    for b, idx in enumerate(batch_indices):
        # Build training index (current batch + buffer)
        train_idx = _to_list(idx)
        if buffer:
            train_idx += buffer.get()
        train_idx = list(dict.fromkeys(train_idx))
        train_idx_t = torch.tensor(train_idx, dtype=torch.long, device=device)

        # Train this batch
        for epoch in range(epochs_per_batch):
            model.train()
            logits = model(graph, features)
            loss = loss_fn(logits[train_idx_t], labels[train_idx_t])
            if ewc: loss = loss + ewc.penalty(model)
            if si:
                grads = torch.autograd.grad(loss, model.parameters(), retain_graph=True, allow_unused=True)
                si.update(grads)
            opt.zero_grad(); loss.backward(); opt.step()
        if si: si.consolidate()
        if ewc: ewc.compute_fisher(graph, features, labels, train_idx)

        # Update rehearsal buffer
        if buffer: buffer.add(idx)

        # EVALUATION: only on samples from classes seen so far
        seen_classes = [c for batch in class_batches[:b+1] for c in batch]
        seen_class_ids = [class_to_label[c] for c in seen_classes]
        class_names_seen = [label_encoder.classes_[i] for i in seen_class_ids]

        # Build evaluation index mask
        mask_seen = np.isin(all_labels_np, seen_class_ids)
        eval_idx = np.where(mask_seen)[0]
        eval_idx_t = torch.tensor(eval_idx, dtype=torch.long, device=device)

        # Build contiguous mapping for labels: old_id -> [0..K-1]
        id_to_seen = {cid: i for i, cid in enumerate(seen_class_ids)}
        y_true = labels[eval_idx_t].cpu().numpy()
        y_true_mapped = np.vectorize(id_to_seen.get)(y_true).astype(np.int64)

        # Slice logits to seen class columns before argmax
        with torch.no_grad():
            logits = model(graph, features)  # shape [N, 8]
            # Gather seen class columns
            seen_cols_t = torch.tensor(seen_class_ids, dtype=torch.long, device=device)
            logits_seen = logits[eval_idx_t][:, seen_cols_t]  # [n_eval, K_seen]
            y_pred_mapped = logits_seen.argmax(dim=1).cpu().numpy()  # already 0..K_seen-1

        acc, cm, pr, rc, f1 = compute_metrics(y_true_mapped, y_pred_mapped, class_names_seen)
        print_batch_report(mech, b+1, class_batches[b], acc, cm, pr, rc, f1, class_names_seen)

        results["acc"].append(acc); results["prec"].append(pr)
        results["rec"].append(rc); results["f1"].append(f1); results["cm"].append(cm)

    return results

# -------------------------
# Run all mechanisms
# -------------------------
mechanisms = ["naive","rehearsal","ewc","si","gen_replay"]
results_all = {}
for mech in mechanisms:
    print(f"\nRunning mechanism: {mech}")
    run_mech = "rehearsal" if mech=="gen_replay" else mech
    res = run_experiment(mech=run_mech, epochs_per_batch=5, lr=1e-3)
    results_all[mech] = res

# -------------------------
# Consolidated summary table
# -------------------------
rows = []
for batch_idx in range(len(class_batches)):
    row = {"Batch": batch_idx + 1, "Classes": ", ".join(class_batches[batch_idx])}
    for mech, res in results_all.items():
        row[f"{mech}_acc"]  = res["acc"][batch_idx]
        row[f"{mech}_prec"] = res["prec"][batch_idx]
        row[f"{mech}_rec"]  = res["rec"][batch_idx]
        row[f"{mech}_f1"]   = res["f1"][batch_idx]
    rows.append(row)

df_summary = pd.DataFrame(rows).round(3)
print("\n=== Consolidated Metrics (cumulative evaluation over seen classes at each batch) ===")
print(tabulate(df_summary, headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))
```

#### Applying some hyper-parameter tuning to check the model performance outcome

This end-to-end script integrates the tuned hyperparameters:
- Hidden dimension 128 with dropout 0.5
- Adam with learning rate 5e-4 and weight decay 1e-4
- More epochs per batch (15)
- Stronger regularization for mechanisms: EWC ?=1000, SI c=5.0
- Larger rehearsal buffer (5000)


```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import dgl
import dgl.nn as dglnn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_fscore_support
from tabulate import tabulate

# -------------------------
# Load and Prepare Dataset
# -------------------------
df = pd.read_csv("all_circuits_features.csv")

# Label encoding
label_encoder = LabelEncoder()
df["gate_label"] = label_encoder.fit_transform(df["gate_type"])

# Feature columns
feature_columns = [
    "fan_in", "fan_out", "dist_to_output", "is_primary_input", "is_primary_output",
    "is_internal", "is_key_gate", "degree_centrality", "betweenness_centrality",
    "closeness_centrality", "clustering_coefficient", "avg_fan_in_neighbors", "avg_fan_out_neighbors"
]

# Clean and normalize
df = df.dropna(subset=feature_columns)
df[feature_columns] = df[feature_columns].astype(float)
scaler = StandardScaler()
df[feature_columns] = scaler.fit_transform(df[feature_columns])

# -------------------------
# Build graph from CSV (simple fan-in heuristic)
# -------------------------
nodes = df["node"].tolist()
node_to_id = {node: i for i, node in enumerate(nodes)}

edges = []
for _, row in df.iterrows():
    node_id = node_to_id[row["node"]]
    potential_sources = df[df["fan_out"] > 0]["node"].tolist()
    k = int(row["fan_in"]) if not np.isnan(row["fan_in"]) else 0
    sources = potential_sources[:max(k, 0)]
    for src in sources:
        if src in node_to_id:
            edges.append((node_to_id[src], node_id))

src_nodes, dst_nodes = zip(*edges) if edges else ([], [])
src_tensor = torch.tensor(src_nodes, dtype=torch.int64)
dst_tensor = torch.tensor(dst_nodes, dtype=torch.int64)
valid_edges = (src_tensor >= 0) & (dst_tensor >= 0) & (src_tensor < len(nodes)) & (dst_tensor < len(nodes))
src_tensor = src_tensor[valid_edges]
dst_tensor = dst_tensor[valid_edges]

graph = dgl.graph((src_tensor, dst_tensor), num_nodes=len(nodes))
graph = dgl.add_self_loop(graph)
graph.ndata['features'] = torch.tensor(df[feature_columns].values, dtype=torch.float32)
graph.ndata['labels'] = torch.tensor(df["gate_label"].values, dtype=torch.long)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
graph = graph.to(device)
graph.ndata['features'] = graph.ndata['features'].to(device)
graph.ndata['labels'] = graph.ndata['labels'].to(device)

# -------------------------
# Class-incremental batches
# -------------------------
class_to_label = {c: i for i, c in enumerate(label_encoder.classes_)}
class_batches = [
    ["and", "not"],   # Batch 1
    ["nor"],          # Batch 2
    ["nand"],         # Batch 3
    ["input"],        # Batch 4
    ["output", "or"], # Batch 5
    ["xor"]           # Batch 6
]

all_labels_np = graph.ndata['labels'].cpu().numpy()

def indices_for_classes(class_names):
    class_ids = [class_to_label[c] for c in class_names]
    mask = np.isin(all_labels_np, class_ids)
    return np.where(mask)[0]

batch_indices = [indices_for_classes(b) for b in class_batches]

# -------------------------
# Tuned GCN model (static 8-way)
# -------------------------
class GCN(nn.Module):
    def __init__(self, in_feats, hidden_feats, out_feats, dropout=0.5):
        super(GCN, self).__init__()
        self.conv1 = dglnn.GraphConv(in_feats, hidden_feats, allow_zero_in_degree=True)
        self.conv2 = dglnn.GraphConv(hidden_feats, out_feats, allow_zero_in_degree=True)
        self.dropout = nn.Dropout(dropout)
    def forward(self, g, inputs):
        x = self.conv1(g, inputs)
        x = torch.relu(x)
        x = self.dropout(x)
        x = self.conv2(g, x)
        return x

in_feats = graph.ndata['features'].shape[1]
hidden_feats = 128
out_feats = len(label_encoder.classes_)

# -------------------------
# Mechanisms
# -------------------------
def _to_list(idx):
    if isinstance(idx, np.ndarray): return idx.tolist()
    if isinstance(idx, list): return idx
    return list(idx)

class RehearsalBuffer:
    def __init__(self, size=5000, num_nodes=None):
        self.size = size
        self.data_idx = []
        self.num_nodes = num_nodes
    def add(self, idx):
        idx_list = _to_list(idx)
        self.data_idx.extend(idx_list)
        if self.num_nodes is not None:
            self.data_idx = [i for i in self.data_idx if 0 <= i < self.num_nodes]
        self.data_idx = list(dict.fromkeys(self.data_idx))
        if len(self.data_idx) > self.size:
            self.data_idx = np.random.choice(self.data_idx, self.size, replace=False).tolist()
    def get(self):
        return self.data_idx

class EWC:
    def __init__(self, model, lam=1000.0):
        self.model = model
        self.lam = lam
        self.star_vars = None
        self.fisher = None
    def compute_fisher(self, g, features, labels, idx, sample_size=3000):
        idx_list = _to_list(idx)
        if len(idx_list) == 0: return
        sample_idx = np.random.choice(idx_list, min(sample_size, len(idx_list)), replace=False)
        sample_idx_t = torch.tensor(sample_idx, dtype=torch.long, device=device)
        self.model.train()
        logits = self.model(g, features)
        loss = nn.CrossEntropyLoss()(logits[sample_idx_t], labels[sample_idx_t])
        self.model.zero_grad()
        loss.backward()
        self.fisher = [p.grad.detach().cpu().numpy()**2 if p.grad is not None else np.zeros_like(p.detach().cpu().numpy())
                       for p in self.model.parameters()]
        self.star_vars = [p.detach().cpu().numpy().copy() for p in self.model.parameters()]
    def penalty(self, model):
        if self.fisher is None:
            return torch.tensor(0.0, device=device)
        total = 0.0
        for p, p_star, f in zip(model.parameters(), self.star_vars, self.fisher):
            total += np.sum(f * (p.detach().cpu().numpy() - p_star)**2)
        return torch.tensor((self.lam/2.0)*total, dtype=torch.float32, device=device)

class SITracker:
    def __init__(self, model, c=5.0, eps=1e-3):
        self.model = model
        self.c = c
        self.eps = eps
        self.prev_vars = [p.detach().cpu().numpy().copy() for p in model.parameters()]
        self.w_importance = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
        self.path_int = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
    def update(self, grads):
        for i,(g,p) in enumerate(zip(grads,self.model.parameters())):
            if g is not None:
                delta = p.detach().cpu().numpy() - self.prev_vars[i]
                self.path_int[i] += (-g.detach().cpu().numpy()) * delta
            self.prev_vars[i] = p.detach().cpu().numpy().copy()
    def consolidate(self):
        for i,p in enumerate(self.model.parameters()):
            delta = p.detach().cpu().numpy() - self.prev_vars[i]
            self.w_importance[i] += self.path_int[i]/(delta**2+self.eps)
            self.path_int[i] = np.zeros_like(p.detach().cpu().numpy())
    def penalty(self, model):
        total=0.0
        for i,p in enumerate(model.parameters()):
            total += np.sum(self.w_importance[i]*(p.detach().cpu().numpy()-self.prev_vars[i])**2)
        return torch.tensor(self.c*total, dtype=torch.float32, device=device)

# -------------------------
# Metrics helpers
# -------------------------
def compute_metrics(y_true, y_pred, class_names):
    labels = list(range(len(class_names)))
    acc = accuracy_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    pr, rc, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, average="macro", zero_division=0
    )
    return acc, cm, pr, rc, f1

def print_batch_report(mech, batch_num, batch_classes, acc, cm, pr, rc, f1, class_names):
    print(f"\n=== Mechanism: {mech} | After batch {batch_num} ({batch_classes}) ===")
    print(f"Accuracy: {acc:.4f}, Precision (macro): {pr:.4f}, Recall: {rc:.4f}, F1 (macro): {f1:.4f}")
    print(pd.DataFrame(cm, index=class_names, columns=class_names))

# -------------------------
# Training loop per mechanism
# -------------------------
def run_experiment(mech="naive", epochs_per_batch=15, lr=5e-4, weight_decay=1e-4):
    model = GCN(in_feats, hidden_feats, out_feats, dropout=0.5).to(device)
    opt = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    loss_fn = nn.CrossEntropyLoss()

    buffer = RehearsalBuffer(5000, num_nodes=graph.num_nodes()) if mech in ["rehearsal","gen_replay"] else None
    ewc = EWC(model, lam=1000.0) if mech=="ewc" else None
    si = SITracker(model, c=5.0) if mech=="si" else None

    results = {"acc":[], "prec":[], "rec":[], "f1":[], "cm":[]}

    features = graph.ndata['features']
    labels = graph.ndata['labels']

    for b, idx in enumerate(batch_indices):
        # Build training index (current batch + buffer)
        train_idx = _to_list(idx)
        if buffer:
            train_idx += buffer.get()
        train_idx = list(dict.fromkeys(train_idx))
        train_idx_t = torch.tensor(train_idx, dtype=torch.long, device=device)

        # Train this batch
        for epoch in range(epochs_per_batch):
            model.train()
            logits = model(graph, features)
            loss = loss_fn(logits[train_idx_t], labels[train_idx_t])
            if ewc: loss = loss + ewc.penalty(model)
            if si:
                grads = torch.autograd.grad(loss, model.parameters(), retain_graph=True, allow_unused=True)
                si.update(grads)
            opt.zero_grad(); loss.backward(); opt.step()
        if si: si.consolidate()
        if ewc: ewc.compute_fisher(graph, features, labels, train_idx)

        # Update rehearsal buffer
        if buffer: buffer.add(idx)

        # EVALUATION: only on samples from classes seen so far
        seen_classes = [c for batch in class_batches[:b+1] for c in batch]
        seen_class_ids = [class_to_label[c] for c in seen_classes]
        class_names_seen = [label_encoder.classes_[i] for i in seen_class_ids]

        mask_seen = np.isin(all_labels_np, seen_class_ids)
        eval_idx = np.where(mask_seen)[0]
        eval_idx_t = torch.tensor(eval_idx, dtype=torch.long, device=device)

        id_to_seen = {cid: i for i, cid in enumerate(seen_class_ids)}
        y_true = labels[eval_idx_t].cpu().numpy()
        y_true_mapped = np.vectorize(id_to_seen.get)(y_true).astype(np.int64)

        with torch.no_grad():
            logits = model(graph, features)  # [N, 8]
            seen_cols_t = torch.tensor(seen_class_ids, dtype=torch.long, device=device)
            logits_seen = logits[eval_idx_t][:, seen_cols_t]  # [n_eval, K_seen]
            y_pred_mapped = logits_seen.argmax(dim=1).cpu().numpy()

        acc, cm, pr, rc, f1 = compute_metrics(y_true_mapped, y_pred_mapped, class_names_seen)
        print_batch_report(mech, b+1, class_batches[b], acc, cm, pr, rc, f1, class_names_seen)

        results["acc"].append(acc); results["prec"].append(pr)
        results["rec"].append(rc); results["f1"].append(f1); results["cm"].append(cm)

    return results

# -------------------------
# Run all mechanisms
# -------------------------
mechanisms = ["naive","rehearsal","ewc","si","gen_replay"]
results_all = {}
for mech in mechanisms:
    print(f"\nRunning mechanism: {mech}")
    run_mech = "rehearsal" if mech=="gen_replay" else mech
    res = run_experiment(mech=run_mech, epochs_per_batch=15, lr=5e-4, weight_decay=1e-4)
    results_all[mech] = res

# -------------------------
# Consolidated summary table
# -------------------------
rows = []
for batch_idx in range(len(class_batches)):
    row = {"Batch": batch_idx + 1, "Classes": ", ".join(class_batches[batch_idx])}
    for mech, res in results_all.items():
        row[f"{mech}_acc"]  = res["acc"][batch_idx]
        row[f"{mech}_prec"] = res["prec"][batch_idx]
        row[f"{mech}_rec"]  = res["rec"][batch_idx]
        row[f"{mech}_f1"]   = res["f1"][batch_idx]
    rows.append(row)

df_summary = pd.DataFrame(rows).round(3)
print("\n=== Consolidated Metrics (cumulative evaluation over seen classes at each batch) ===")
print(tabulate(df_summary, headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))
```

- Random Search
- Bayesian Optimization (surrogate + exploration)
- Hyperband (successive halving)
- Population-Based Training (PBT)
- Evolutionary Algorithm (GA)

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import dgl
import dgl.nn as dglnn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_fscore_support, classification_report
from tabulate import tabulate
import random
import math
import time

# -------------------------
# Load and Prepare Dataset
# -------------------------
df = pd.read_csv("all_circuits_features.csv")

label_encoder = LabelEncoder()
df["gate_label"] = label_encoder.fit_transform(df["gate_type"])

feature_columns = [
    "fan_in", "fan_out", "dist_to_output", "is_primary_input", "is_primary_output",
    "is_internal", "is_key_gate", "degree_centrality", "betweenness_centrality",
    "closeness_centrality", "clustering_coefficient", "avg_fan_in_neighbors", "avg_fan_out_neighbors"
]
df = df.dropna(subset=feature_columns)
df[feature_columns] = df[feature_columns].astype(float)
scaler = StandardScaler()
df[feature_columns] = scaler.fit_transform(df[feature_columns])

# -------------------------
# Build graph
# -------------------------
nodes = df["node"].tolist()
node_to_id = {node: i for i, node in enumerate(nodes)}
edges = []
for _, row in df.iterrows():
    node_id = node_to_id[row["node"]]
    potential_sources = df[df["fan_out"] > 0]["node"].tolist()
    k = int(row["fan_in"]) if not np.isnan(row["fan_in"]) else 0
    sources = potential_sources[:max(k, 0)]
    for src in sources:
        if src in node_to_id:
            edges.append((node_to_id[src], node_id))

src_nodes, dst_nodes = zip(*edges) if edges else ([], [])
src_tensor = torch.tensor(src_nodes, dtype=torch.int64)
dst_tensor = torch.tensor(dst_nodes, dtype=torch.int64)
valid_edges = (src_tensor >= 0) & (dst_tensor >= 0) & (src_tensor < len(nodes)) & (dst_tensor < len(nodes))
src_tensor = src_tensor[valid_edges]
dst_tensor = dst_tensor[valid_edges]

graph = dgl.graph((src_tensor, dst_tensor), num_nodes=len(nodes))
graph = dgl.add_self_loop(graph)
graph.ndata['features'] = torch.tensor(df[feature_columns].values, dtype=torch.float32)
graph.ndata['labels'] = torch.tensor(df["gate_label"].values, dtype=torch.long)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
graph = graph.to(device)
graph.ndata['features'] = graph.ndata['features'].to(device)
graph.ndata['labels'] = graph.ndata['labels'].to(device)

# -------------------------
# Class-incremental batches
# -------------------------
class_to_label = {c: i for i, c in enumerate(label_encoder.classes_)}
class_batches = [
    ["and", "not"],
    ["nor"],
    ["nand"],
    ["input"],
    ["output", "or"],
    ["xor"]
]
all_labels_np = graph.ndata['labels'].cpu().numpy()

def indices_for_classes(class_names):
    class_ids = [class_to_label[c] for c in class_names]
    mask = np.isin(all_labels_np, class_ids)
    return np.where(mask)[0]
batch_indices = [indices_for_classes(b) for b in class_batches]

# -------------------------
# Model
# -------------------------
class GCN(nn.Module):
    def __init__(self, in_feats, hidden_feats, out_feats, dropout=0.5):
        super(GCN, self).__init__()
        self.conv1 = dglnn.GraphConv(in_feats, hidden_feats, allow_zero_in_degree=True)
        self.conv2 = dglnn.GraphConv(hidden_feats, out_feats, allow_zero_in_degree=True)
        self.dropout = nn.Dropout(dropout)
    def forward(self, g, inputs):
        x = self.conv1(g, inputs)
        x = torch.relu(x)
        x = self.dropout(x)
        x = self.conv2(g, x)
        return x

in_feats = graph.ndata['features'].shape[1]
out_feats = len(label_encoder.classes_)

# -------------------------
# Mechanisms
# -------------------------
def _to_list(idx):
    if isinstance(idx, np.ndarray): return idx.tolist()
    if isinstance(idx, list): return idx
    return list(idx)

class RehearsalBuffer:
    def __init__(self, size=5000, num_nodes=None):
        self.size = size
        self.data_idx = []
        self.num_nodes = num_nodes
    def add(self, idx):
        idx_list = _to_list(idx)
        self.data_idx.extend(idx_list)
        if self.num_nodes is not None:
            self.data_idx = [i for i in self.data_idx if 0 <= i < self.num_nodes]
        self.data_idx = list(dict.fromkeys(self.data_idx))
        if len(self.data_idx) > self.size:
            self.data_idx = np.random.choice(self.data_idx, self.size, replace=False).tolist()
    def get(self):
        return self.data_idx

class EWC:
    def __init__(self, model, lam=1000.0):
        self.model = model
        self.lam = lam
        self.star_vars = None
        self.fisher = None
    def compute_fisher(self, g, features, labels, idx, sample_size=3000):
        idx_list = _to_list(idx)
        if len(idx_list) == 0: return
        sample_idx = np.random.choice(idx_list, min(sample_size, len(idx_list)), replace=False)
        sample_idx_t = torch.tensor(sample_idx, dtype=torch.long, device=device)
        self.model.train()
        logits = self.model(g, features)
        loss = nn.CrossEntropyLoss()(logits[sample_idx_t], labels[sample_idx_t])
        self.model.zero_grad()
        loss.backward()
        self.fisher = [p.grad.detach().cpu().numpy()**2 if p.grad is not None else np.zeros_like(p.detach().cpu().numpy())
                       for p in self.model.parameters()]
        self.star_vars = [p.detach().cpu().numpy().copy() for p in self.model.parameters()]
    def penalty(self, model):
        if self.fisher is None:
            return torch.tensor(0.0, device=device)
        total = 0.0
        for p, p_star, f in zip(model.parameters(), self.star_vars, self.fisher):
            total += np.sum(f * (p.detach().cpu().numpy() - p_star)**2)
        return torch.tensor((self.lam/2.0)*total, dtype=torch.float32, device=device)

class SITracker:
    def __init__(self, model, c=5.0, eps=1e-3):
        self.model = model
        self.c = c
        self.eps = eps
        self.prev_vars = [p.detach().cpu().numpy().copy() for p in model.parameters()]
        self.w_importance = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
        self.path_int = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
    def update(self, grads):
        for i,(g,p) in enumerate(zip(grads,self.model.parameters())):
            if g is not None:
                delta = p.detach().cpu().numpy() - self.prev_vars[i]
                self.path_int[i] += (-g.detach().cpu().numpy()) * delta
            self.prev_vars[i] = p.detach().cpu().numpy().copy()
    def consolidate(self):
        for i,p in enumerate(self.model.parameters()):
            delta = p.detach().cpu().numpy() - self.prev_vars[i]
            self.w_importance[i] += self.path_int[i]/(delta**2+self.eps)
            self.path_int[i] = np.zeros_like(p.detach().cpu().numpy())
    def penalty(self, model):
        total=0.0
        for i,p in enumerate(model.parameters()):
            total += np.sum(self.w_importance[i]*(p.detach().cpu().numpy()-self.prev_vars[i])**2)
        return torch.tensor(self.c*total, dtype=torch.float32, device=device)

# -------------------------
# Metrics helpers
# -------------------------
def compute_metrics(y_true, y_pred, class_names):
    labels = list(range(len(class_names)))
    acc = accuracy_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    pr, rc, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, average="macro", zero_division=0
    )
    return acc, cm, pr, rc, f1

def eval_cumulative(model, features, labels, b, class_batches, label_encoder, class_to_label, all_labels_np):
    # Seen classes and contiguous mapping
    seen_classes = [c for batch in class_batches[:b+1] for c in batch]
    seen_class_ids = [class_to_label[c] for c in seen_classes]
    class_names_seen = [label_encoder.classes_[i] for i in seen_class_ids]
    id_to_seen = {cid: i for i, cid in enumerate(seen_class_ids)}

    # Build eval index
    mask_seen = np.isin(all_labels_np, seen_class_ids)
    eval_idx = np.where(mask_seen)[0]
    eval_idx_t = torch.tensor(eval_idx, dtype=torch.long, device=device)

    # Map y_true to contiguous
    y_true = labels[eval_idx_t].cpu().numpy()
    y_true_mapped = np.vectorize(id_to_seen.get)(y_true).astype(np.int64)

    # Slice logits to seen columns
    with torch.no_grad():
        logits = model(graph, features)
        seen_cols_t = torch.tensor(seen_class_ids, dtype=torch.long, device=device)
        logits_seen = logits[eval_idx_t][:, seen_cols_t]
        y_pred_mapped = logits_seen.argmax(dim=1).cpu().numpy()

    # Metrics + confusion matrix
    acc, cm, pr, rc, f1 = compute_metrics(y_true_mapped, y_pred_mapped, class_names_seen)
    # Classification report for seen classes
    report = classification_report(y_true_mapped, y_pred_mapped, target_names=class_names_seen, digits=4)
    return acc, cm, pr, rc, f1, report, class_names_seen

# -------------------------
# Training loop (one run)
# -------------------------
def run_experiment(mech="naive",
                   hidden_feats=128,
                   dropout=0.5,
                   lr=5e-4,
                   weight_decay=1e-4,
                   epochs_per_batch=15,
                   buffer_size=5000,
                   ewc_lam=1000.0,
                   si_c=5.0):
    model = GCN(in_feats, hidden_feats, out_feats, dropout=dropout).to(device)
    opt = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    loss_fn = nn.CrossEntropyLoss()

    buffer = RehearsalBuffer(buffer_size, num_nodes=graph.num_nodes()) if mech in ["rehearsal","gen_replay"] else None
    ewc = EWC(model, lam=ewc_lam) if mech=="ewc" else None
    si = SITracker(model, c=si_c) if mech=="si" else None

    features = graph.ndata['features']
    labels = graph.ndata['labels']

    batch_accs = []
    batch_reports = []
    batch_confusions = []
    batch_classnames = []

    for b, idx in enumerate(batch_indices):
        # Training set for this batch
        train_idx = _to_list(idx)
        if buffer: train_idx += buffer.get()
        train_idx = list(dict.fromkeys(train_idx))
        train_idx_t = torch.tensor(train_idx, dtype=torch.long, device=device)

        # Train
        for epoch in range(epochs_per_batch):
            model.train()
            logits = model(graph, features)
            loss = loss_fn(logits[train_idx_t], labels[train_idx_t])
            if ewc: loss = loss + ewc.penalty(model)
            if si:
                grads = torch.autograd.grad(loss, model.parameters(), retain_graph=True, allow_unused=True)
                si.update(grads)
            opt.zero_grad(); loss.backward(); opt.step()
        if si: si.consolidate()
        if ewc: ewc.compute_fisher(graph, features, labels, train_idx)

        # Update buffer
        if buffer: buffer.add(idx)

        # Evaluate cumulatively on seen classes
        acc, cm, pr, rc, f1, report, class_names_seen = eval_cumulative(
            model, features, labels, b, class_batches, label_encoder, class_to_label, all_labels_np
        )
        batch_accs.append(acc)
        batch_confusions.append(cm)
        batch_reports.append(report)
        batch_classnames.append(class_names_seen)

    return {
        "config": {
            "mech": mech, "hidden_feats": hidden_feats, "dropout": dropout,
            "lr": lr, "weight_decay": weight_decay, "epochs_per_batch": epochs_per_batch,
            "buffer_size": buffer_size, "ewc_lam": ewc_lam, "si_c": si_c
        },
        "batch_accs": batch_accs,
        "batch_reports": batch_reports,
        "batch_confusions": batch_confusions,
        "batch_classnames": batch_classnames
    }

# -------------------------
# Search spaces
# -------------------------
MECHANISMS = ["naive","rehearsal","ewc","si","gen_replay"]
HSPACE = {
    "hidden_feats": [64, 128, 256],
    "dropout": [0.0, 0.3, 0.5],
    "lr": [1e-3, 5e-4, 1e-4],
    "weight_decay": [0.0, 1e-5, 1e-4],
    "epochs_per_batch": [10, 15, 20],
    "buffer_size": [2000, 5000, 10000],
    "ewc_lam": [500.0, 1000.0, 2000.0],
    "si_c": [2.0, 5.0, 10.0],
}

def sample_random():
    return {
        "hidden_feats": random.choice(HSPACE["hidden_feats"]),
        "dropout": random.choice(HSPACE["dropout"]),
        "lr": random.choice(HSPACE["lr"]),
        "weight_decay": random.choice(HSPACE["weight_decay"]),
        "epochs_per_batch": random.choice(HSPACE["epochs_per_batch"]),
        "buffer_size": random.choice(HSPACE["buffer_size"]),
        "ewc_lam": random.choice(HSPACE["ewc_lam"]),
        "si_c": random.choice(HSPACE["si_c"]),
    }

# -------------------------
# Objective: average cumulative accuracy over 6 batches
# -------------------------
def objective(config, mech):
    res = run_experiment(mech=mech, **config)
    avg_acc = float(np.mean(res["batch_accs"]))
    return avg_acc, res

# -------------------------
# 1) Grid Search (coarse)
# -------------------------
def grid_search(max_trials_per_mech=20):
    trials = []
    for mech in MECHANISMS:
        combos = []
        # Build a coarse grid to cap trials
        for hf in HSPACE["hidden_feats"]:
            for dr in [0.3, 0.5]:
                for lr in [1e-3, 5e-4]:
                    for wd in [0.0, 1e-4]:
                        for ep in [10, 15]:
                            cfg = {"hidden_feats": hf, "dropout": dr, "lr": lr,
                                   "weight_decay": wd, "epochs_per_batch": ep,
                                   "buffer_size": 5000, "ewc_lam": 1000.0, "si_c": 5.0}
                            combos.append(cfg)
        combos = combos[:max_trials_per_mech]
        for cfg in combos:
            avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
            trials.append({"method":"grid", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
    return trials

# -------------------------
# 2) Random Search
# -------------------------
def random_search(trials_per_mech=20):
    trials = []
    for mech in MECHANISMS:
        for _ in range(trials_per_mech):
            cfg = sample_random()
            avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
            trials.append({"method":"random", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
    return trials

# -------------------------
# 3) Bayesian Optimization (TPE-like)
# -------------------------
def bayes_opt(trials_per_mech=25):
    trials = []
    history = {m: [] for m in MECHANISMS}
    for mech in MECHANISMS:
        for t in range(trials_per_mech):
            if t < 5:
                cfg = sample_random()
            else:
                # Bias sampling towards promising regions
                past = sorted(history[mech], key=lambda x: -x["avg_acc"])
                top = past[:max(1, len(past)//3)]
                if top:
                    base = random.choice(top)["config"]
                    cfg = {
                        "hidden_feats": random.choice([base["hidden_feats"]] + HSPACE["hidden_feats"]),
                        "dropout": random.choice([base["dropout"], 0.3, 0.5]),
                        "lr": random.choice([base["lr"], 5e-4, 1e-4]),
                        "weight_decay": random.choice([base["weight_decay"], 1e-5, 1e-4]),
                        "epochs_per_batch": random.choice([base["epochs_per_batch"], 15, 20]),
                        "buffer_size": random.choice([base["buffer_size"], 5000, 10000]),
                        "ewc_lam": random.choice([base["ewc_lam"], 1000.0, 2000.0]),
                        "si_c": random.choice([base["si_c"], 5.0, 10.0]),
                    }
                else:
                    cfg = sample_random()
            avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
            record = {"method":"bayes", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res}
            trials.append(record); history[mech].append(record)
    return trials

# -------------------------
# 4) Hyperband (successive halving)
# -------------------------
def hyperband(R=27, eta=3):  # R max epochs, eta=downsampling rate
    # We'll treat epochs_per_batch as the resource
    trials = []
    for mech in MECHANISMS:
        s_max = int(math.log(R, eta))
        for s in range(s_max, -1, -1):
            n = int(math.ceil((s_max+1)/(s+1)) * (eta**s))
            r = int(R * (eta**(-s)))
            # Sample n configs
            configs = [sample_random() for _ in range(n)]
            for cfg in configs:
                cfg["epochs_per_batch"] = r
            # Evaluate and downselect
            records = []
            for cfg in configs:
                avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
                records.append({"method":"hyperband", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
            # Successive halving: keep top 1/eta
            k = max(1, n//eta)
            records = sorted(records, key=lambda x: -x["avg_acc"])[:k]
            trials.extend(records)
    return trials

# -------------------------
# 5) Population-Based Training (PBT)
# -------------------------
def pbt(generations=3, population_size=6, mutate_prob=0.5):
    trials = []
    for mech in MECHANISMS:
        # Init population
        population = []
        for _ in range(population_size):
            cfg = sample_random()
            avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
            population.append({"method":"pbt", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
        trials.extend(population)
        # Evolve
        for _ in range(generations):
            population = sorted(population, key=lambda x: -x["avg_acc"])
            elites = population[:population_size//2]
            offspring = []
            for e in elites:
                cfg = e["config"].copy()
                # Mutate some params
                if random.random() < mutate_prob:
                    cfg["lr"] = random.choice(HSPACE["lr"])
                if random.random() < mutate_prob:
                    cfg["hidden_feats"] = random.choice(HSPACE["hidden_feats"])
                if random.random() < mutate_prob:
                    cfg["dropout"] = random.choice(HSPACE["dropout"])
                if random.random() < mutate_prob:
                    cfg["epochs_per_batch"] = random.choice(HSPACE["epochs_per_batch"])
                if random.random() < mutate_prob:
                    cfg["weight_decay"] = random.choice(HSPACE["weight_decay"])
                if random.random() < mutate_prob and mech in ["rehearsal","gen_replay"]:
                    cfg["buffer_size"] = random.choice(HSPACE["buffer_size"])
                if random.random() < mutate_prob and mech=="ewc":
                    cfg["ewc_lam"] = random.choice(HSPACE["ewc_lam"])
                if random.random() < mutate_prob and mech=="si":
                    cfg["si_c"] = random.choice(HSPACE["si_c"])
                avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
                offspring.append({"method":"pbt", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
            population = elites + offspring
            trials.extend(offspring)
    return trials

# -------------------------
# 6) Genetic Algorithm (GA)
# -------------------------
def ga(generations=3, population_size=8, crossover_rate=0.7, mutation_rate=0.2):
    trials = []
    for mech in MECHANISMS:
        # Initialize population
        pop = []
        for _ in range(population_size):
            cfg = sample_random()
            avg_acc, res = objective(cfg, mech if mech!="gen_replay" else "rehearsal")
            pop.append({"method":"ga", "mech":mech, "config":cfg, "avg_acc":avg_acc, "res":res})
        trials.extend(pop)
        # Evolution loop
        for _ in range(generations):
            pop = sorted(pop, key=lambda x: -x["avg_acc"])
            new_pop = pop[:2]  # elitism
            while len(new_pop) < population_size:
                # Select parents
                p1, p2 = random.choice(pop[:len(pop)//2]), random.choice(pop[len(pop)//2:])
                child_cfg = p1["config"].copy()
                # Crossover
                if random.random() < crossover_rate:
                    for k in child_cfg:
                        if random.random() < 0.5:
                            child_cfg[k] = p2["config"][k]
                # Mutation
                for k in child_cfg:
                    if random.random() < mutation_rate:
                        child_cfg[k] = sample_random()[k]
                avg_acc, res = objective(child_cfg, mech if mech!="gen_replay" else "rehearsal")
                new_pop.append({"method":"ga", "mech":mech, "config":child_cfg, "avg_acc":avg_acc, "res":res})
            pop = new_pop
            trials.extend(new_pop[2:])
    return trials

# -------------------------
# Run all search methods
# -------------------------
start_time = time.time()
results_trials = []
results_trials += grid_search(max_trials_per_mech=12)
results_trials += random_search(trials_per_mech=12)
results_trials += bayes_opt(trials_per_mech=15)
results_trials += hyperband(R=27, eta=3)
results_trials += pbt(generations=2, population_size=6)
results_trials += ga(generations=2, population_size=8)
elapsed = time.time() - start_time
print(f"\nSearch completed in {elapsed/60:.1f} minutes. Total trials: {len(results_trials)}")

# -------------------------
# Consolidated reporting
# -------------------------
# Flatten per-trial summary rows (avg accuracy across batches and final batch report)
summary_rows = []
for t in results_trials:
    res = t["res"]
    config = res["config"].copy()
    config.update({"method": t["method"], "mech": t["mech"]})
    row = {
        **config,
        "avg_acc": float(np.mean(res["batch_accs"])),
        "batch1_acc": res["batch_accs"][0],
        "batch2_acc": res["batch_accs"][1],
        "batch3_acc": res["batch_accs"][2],
        "batch4_acc": res["batch_accs"][3],
        "batch5_acc": res["batch_accs"][4],
        "batch6_acc": res["batch_accs"][5],
    }
    summary_rows.append(row)

df_summary = pd.DataFrame(summary_rows)

# Pretty print top results per method
print("\n=== Top-5 configurations per search method (by avg cumulative accuracy) ===")
for method in ["grid","random","bayes","hyperband","pbt","ga"]:
    df_top = df_summary[df_summary["method"]==method].sort_values("avg_acc", ascending=False).head(5)
    print(f"\n[{method.upper()}] Top-5")
    print(tabulate(df_top.round(3), headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))

# Overall top configurations
df_overall_top = df_summary.sort_values("avg_acc", ascending=False).head(10)
print("\n=== Overall Top-10 configurations ===")
print(tabulate(df_overall_top.round(3), headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))

# -------------------------
# Detailed reports for the best config per mechanism
# -------------------------
print("\n=== Detailed classification reports for best config per mechanism ===")
for mech in MECHANISMS:
    # among all trials with this mech, pick best avg_acc
    mech_trials = [t for t in results_trials if t["mech"]==mech]
    if not mech_trials: continue
    best = max(mech_trials, key=lambda x: x["avg_acc"])
    res = best["res"]
    cfg = best["config"]
    print(f"\nMechanism: {mech} | Method: {best['method']} | AvgAcc: {np.mean(res['batch_accs']):.3f}")
    print(f"Config: {cfg}")

    # Show batch-wise metrics and classification report
    for b in range(len(class_batches)):
        classes_seen = ", ".join([c for batch in class_batches[:b+1] for c in batch])
        print(f"\n- Batch {b+1} (seen: {classes_seen}) | Acc: {res['batch_accs'][b]:.3f}")
        print("Confusion matrix:")
        cm_df = pd.DataFrame(res["batch_confusions"][b],
                             index=res["batch_classnames"][b],
                             columns=res["batch_classnames"][b])
        print(cm_df)
        print("\nClassification report:")
        print(res["batch_reports"][b])
```

#### Dynamic Architecture

Below is a direct adaptation to a dynamic architecture like MNIST example: the final layer expands per batch to add new class columns while preserving old weights. It preserves all mechanisms (Naive, Rehearsal, EWC, SI, Gen. Replay) and plugs into six search strategies with minimal changes.

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import dgl
import dgl.nn as dglnn
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_fscore_support, classification_report
from tabulate import tabulate
import random, math, time, copy
from tqdm.notebook import tqdm

# -------------------------
# Load and Prepare Dataset
# -------------------------
df = pd.read_csv("all_circuits_features.csv")

label_encoder = LabelEncoder()
df["gate_label"] = label_encoder.fit_transform(df["gate_type"])

feature_columns = [
    "fan_in", "fan_out", "dist_to_output", "is_primary_input", "is_primary_output",
    "is_internal", "is_key_gate", "degree_centrality", "betweenness_centrality",
    "closeness_centrality", "clustering_coefficient", "avg_fan_in_neighbors", "avg_fan_out_neighbors"
]
df = df.dropna(subset=feature_columns)
df[feature_columns] = df[feature_columns].astype(float)
scaler = StandardScaler()
df[feature_columns] = scaler.fit_transform(df[feature_columns])

# -------------------------
# Build graph
# -------------------------
nodes = df["node"].tolist()
node_to_id = {node: i for i, node in enumerate(nodes)}
edges = []
for _, row in df.iterrows():
    node_id = node_to_id[row["node"]]
    potential_sources = df[df["fan_out"] > 0]["node"].tolist()
    k = int(row["fan_in"]) if not np.isnan(row["fan_in"]) else 0
    sources = potential_sources[:max(k, 0)]
    for src in sources:
        if src in node_to_id:
            edges.append((node_to_id[src], node_id))

src_nodes, dst_nodes = zip(*edges) if edges else ([], [])
src_tensor = torch.tensor(src_nodes, dtype=torch.int64)
dst_tensor = torch.tensor(dst_nodes, dtype=torch.int64)
valid_edges = (src_tensor >= 0) & (dst_tensor >= 0) & (src_tensor < len(nodes)) & (dst_tensor < len(nodes))
src_tensor = src_tensor[valid_edges]
dst_tensor = dst_tensor[valid_edges]

graph = dgl.graph((src_tensor, dst_tensor), num_nodes=len(nodes))
graph = dgl.add_self_loop(graph)
graph.ndata['features'] = torch.tensor(df[feature_columns].values, dtype=torch.float32)
graph.ndata['labels'] = torch.tensor(df["gate_label"].values, dtype=torch.long)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
graph = graph.to(device)
graph.ndata['features'] = graph.ndata['features'].to(device)
graph.ndata['labels'] = graph.ndata['labels'].to(device)

# -------------------------
# Class-incremental batches
# -------------------------
class_to_label = {c: i for i, c in enumerate(label_encoder.classes_)}
class_batches = [
    ["and", "not"],
    ["nor"],
    ["nand"],
    ["input"],
    ["output", "or"],
    ["xor"]
]
all_labels_np = graph.ndata['labels'].cpu().numpy()

def indices_for_classes(class_names):
    class_ids = [class_to_label[c] for c in class_names]
    mask = np.isin(all_labels_np, class_ids)
    return np.where(mask)[0]

batch_indices = [indices_for_classes(b) for b in class_batches]

# -------------------------
# Dynamic GCN model (expandable output head)
# -------------------------
class GCN(nn.Module):
    def __init__(self, in_feats, hidden_feats, out_feats, dropout=0.5):
        super(GCN, self).__init__()
        self.conv1 = dglnn.GraphConv(in_feats, hidden_feats, allow_zero_in_degree=True)
        self.conv2 = dglnn.GraphConv(hidden_feats, hidden_feats, allow_zero_in_degree=True)
        self.dropout = nn.Dropout(dropout)
        self.out = nn.Linear(hidden_feats, out_feats)

    def forward(self, g, inputs):
        x = self.conv1(g, inputs)
        x = torch.relu(x)
        x = self.dropout(x)
        x = self.conv2(g, x)
        x = torch.relu(x)
        return self.out(x)

def expand_output_layer(model, new_out_feats, device):
    old_out = model.out
    old_out_feats = old_out.out_features
    if new_out_feats <= old_out_feats:
        return model, 0, 0
    in_features = old_out.in_features
    new_out = nn.Linear(in_features, new_out_feats, bias=True).to(device)
    with torch.no_grad():
        nn.init.normal_(new_out.weight, mean=0.0, std=0.01)
        nn.init.zeros_(new_out.bias)
        # Copy rows (weights shape: [out_features, in_features])
        new_out.weight[:old_out_feats, :] = old_out.weight
        new_out.bias[:old_out_feats] = old_out.bias
    model.out = new_out
    added_units = new_out_feats - old_out_feats
    added_params = added_units * (in_features + 1)
    return model, added_units, added_params

# -------------------------
# Mechanisms (Rehearsal, EWC, SI)
# -------------------------
def _to_list(idx):
    if isinstance(idx, np.ndarray): return idx.tolist()
    if isinstance(idx, list): return idx
    return list(idx)

class RehearsalBuffer:
    def __init__(self, size=5000, num_nodes=None):
        self.size = size
        self.data_idx = []
        self.num_nodes = num_nodes
    def add(self, idx):
        idx_list = _to_list(idx)
        self.data_idx.extend(idx_list)
        if self.num_nodes is not None:
            self.data_idx = [i for i in self.data_idx if 0 <= i < self.num_nodes]
        self.data_idx = list(dict.fromkeys(self.data_idx))
        if len(self.data_idx) > self.size:
            self.data_idx = np.random.choice(self.data_idx, self.size, replace=False).tolist()
    def get(self):
        return self.data_idx

class EWC:
    def __init__(self, model, lam=1000.0):
        self.model = model; self.lam = lam
        self.star_vars = None; self.fisher = None
    def compute_fisher(self, g, features, labels, idx, sample_size=3000):
        idx_list = _to_list(idx)
        if len(idx_list) == 0: return
        sample_idx = np.random.choice(idx_list, min(sample_size, len(idx_list)), replace=False)
        sample_idx_t = torch.tensor(sample_idx, dtype=torch.long, device=device)
        self.model.train()
        logits = self.model(g, features)
        loss = nn.CrossEntropyLoss()(logits[sample_idx_t], labels[sample_idx_t])
        self.model.zero_grad(); loss.backward()
        self.fisher = [p.grad.detach().cpu().numpy()**2 if p.grad is not None else np.zeros_like(p.detach().cpu().numpy())
                       for p in self.model.parameters()]
        self.star_vars = [p.detach().cpu().numpy().copy() for p in self.model.parameters()]
    def penalty(self, model):
        # If snapshots missing or mismatched due to expansion, skip
        if self.fisher is None or self.star_vars is None:
            return torch.tensor(0.0, device=device)
        total = 0.0
        for p, p_star, f in zip(model.parameters(), self.star_vars, self.fisher):
            W = p.detach().cpu().numpy()
            # Align shapes on overlap
            shape = tuple(min(a, b) for a, b in zip(W.shape, f.shape))
            if W.shape != f.shape or W.shape != p_star.shape:
                # slice to common shape
                slices = tuple(slice(0, s) for s in shape)
                W_s = W[slices]; p_star_s = p_star[slices]; f_s = f[slices]
                total += np.sum(f_s * (W_s - p_star_s) ** 2)
            else:
                total += np.sum(f * (W - p_star) ** 2)
        return torch.tensor((self.lam/2.0)*total, dtype=torch.float32, device=device)

class SITracker:
    def __init__(self, model, c=5.0, eps=1e-3):
        self.model = model; self.c = c; self.eps = eps
        self.prev_vars = [p.detach().cpu().numpy().copy() for p in model.parameters()]
        self.w_importance = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
        self.path_int = [np.zeros_like(p.detach().cpu().numpy()) for p in model.parameters()]
    def update(self, grads):
        for i,(g,p) in enumerate(zip(grads,self.model.parameters())):
            if g is not None:
                delta = p.detach().cpu().numpy() - self.prev_vars[i]
                self.path_int[i] += (-g.detach().cpu().numpy()) * delta
            self.prev_vars[i] = p.detach().cpu().numpy().copy()
    def consolidate(self):
        for i,p in enumerate(self.model.parameters()):
            delta = p.detach().cpu().numpy() - self.prev_vars[i]
            self.w_importance[i] += self.path_int[i]/(delta**2+self.eps)
            self.path_int[i] = np.zeros_like(p.detach().cpu().numpy())
    def penalty(self, model):
        total=0.0
        for i,p in enumerate(model.parameters()):
            # align shapes on overlap to be safe if expanded
            W = p.detach().cpu().numpy()
            prev = self.prev_vars[i]
            imp = self.w_importance[i]
            shape = tuple(min(a, b) for a, b in zip(W.shape, prev.shape))
            slices = tuple(slice(0, s) for s in shape)
            W_s = W[slices]; prev_s = prev[slices]; imp_s = imp[slices]
            total += np.sum(imp_s*(W_s - prev_s)**2)
        return torch.tensor(self.c*total, dtype=torch.float32, device=device)

# -------------------------
# Distillation (LwF)
# -------------------------
def soft_targets_probs(model, g, x, temperature=2.0):
    model.eval()
    with torch.no_grad():
        logits = model(g, x)
        probs = torch.softmax(logits, dim=1).clamp(1e-7, 1.0)
        return torch.softmax(torch.log(probs)/temperature, dim=1)

def distillation_kl(student_probs, teacher_soft, temperature=2.0):
    student_probs = student_probs.clamp(1e-7, 1.0)
    s_soft = torch.softmax(torch.log(student_probs)/temperature, dim=1)
    return torch.sum(teacher_soft*(torch.log(teacher_soft+1e-7)-torch.log(s_soft+1e-7)),dim=1).mean()

# -------------------------
# Metrics and evaluation on seen classes only
# -------------------------
def compute_metrics(y_true, y_pred, class_names):
    labels = list(range(len(class_names)))
    acc = accuracy_score(y_true, y_pred)
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    pr, rc, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, labels=labels, average="macro", zero_division=0
    )
    return acc, cm, pr, rc, f1

def eval_cumulative_dynamic(model, features, labels, b, class_batches, label_encoder, class_to_label, all_labels_np):
    seen_classes = [c for batch in class_batches[:b+1] for c in batch]
    seen_class_ids = [class_to_label[c] for c in seen_classes]
    class_names_seen = [label_encoder.classes_[i] for i in seen_class_ids]
    id_to_seen = {cid: i for i,cid in enumerate(seen_class_ids)}

    mask_seen = np.isin(all_labels_np, seen_class_ids)
    eval_idx = np.where(mask_seen)[0]
    eval_idx_t = torch.tensor(eval_idx, dtype=torch.long, device=device)

    y_true_global = labels[eval_idx_t].cpu().numpy()
    y_true_mapped = np.vectorize(id_to_seen.get)(y_true_global).astype(np.int64)

    model.eval()
    with torch.no_grad():
        logits_all = model(graph, features)
        seen_cols = torch.tensor(seen_class_ids, dtype=torch.long, device=device)
        logits_seen = logits_all[eval_idx_t][:, seen_cols]
        y_pred_mapped = logits_seen.argmax(dim=1).cpu().numpy()

    acc, cm, pr, rc, f1 = compute_metrics(y_true_mapped, y_pred_mapped, class_names_seen)
    report = classification_report(y_true_mapped, y_pred_mapped, target_names=class_names_seen, digits=4, zero_division=0)
    return acc, cm, pr, rc, f1, report, class_names_seen

# -------------------------
# Dynamic training loop
# -------------------------
in_feats = graph.ndata['features'].shape[1]

def run_experiment_dynamic(mech="naive",
                           hidden_feats=128,
                           dropout=0.5,
                           lrs=(1e-3,)*6,
                           weight_decay=1e-4,
                           epochs_per_batch=(10,)*6,
                           buffer_size=5000,
                           ewc_lam=1000.0,
                           si_c=5.0,
                           use_distill_for=("ewc","si"),
                           distill_T=2.0,
                           distill_alpha=0.5):
    # Initialize head to cover global IDs in first batch
    first_class_ids = [class_to_label[c] for c in class_batches[0]]
    init_out_feats = max(first_class_ids) + 1
    model = GCN(in_feats, hidden_feats, init_out_feats, dropout=dropout).to(device)
    opt = optim.Adam(model.parameters(), lr=lrs[0], weight_decay=weight_decay)
    loss_fn = nn.CrossEntropyLoss()

    buffer = RehearsalBuffer(buffer_size, num_nodes=graph.num_nodes()) if mech in ["rehearsal","gen_replay"] else None
    ewc = EWC(model, lam=ewc_lam) if mech=="ewc" else None
    si = SITracker(model, c=si_c) if mech=="si" else None

    features, labels = graph.ndata['features'], graph.ndata['labels']
    batch_accs, batch_reports, batch_confusions, batch_classnames = [], [], [], []
    teacher_model = None

    for b, idx in enumerate(batch_indices):
        if mech in use_distill_for:
            teacher_model = copy.deepcopy(model).to(device).eval()

        seen_classes = [c for batch in class_batches[:b+1] for c in batch]
        seen_class_ids = [class_to_label[c] for c in seen_classes]
        target_out = max(seen_class_ids) + 1

        # Expand output head if needed
        if model.out.out_features < target_out:
            model, _, _ = expand_output_layer(model, target_out, device)
            opt = optim.Adam(model.parameters(), lr=lrs[b], weight_decay=weight_decay)
            # Reset regularizers to avoid shape mismatch
            if isinstance(ewc, EWC):
                ewc.model = model
                ewc.star_vars = None
                ewc.fisher = None
            if isinstance(si, SITracker):
                si = SITracker(model, c=si.c)

        train_idx = _to_list(idx)
        if buffer: train_idx += buffer.get()
        train_idx = list(dict.fromkeys(train_idx))
        train_idx_t = torch.tensor(train_idx, dtype=torch.long, device=device)

        # Build mapping from global IDs -> contiguous seen indices for loss targets
        id_to_seen = {cid: i for i, cid in enumerate(seen_class_ids)}
        max_label_id = int(max(seen_class_ids))
        map_tensor = torch.full((max_label_id + 1,), -1, dtype=torch.long, device=device)
        for cid, j in id_to_seen.items():
            map_tensor[cid] = j
        y_train_global = labels[train_idx_t]
        y_seen_mapped = map_tensor[y_train_global]  # in [0, C_seen-1]

        for epoch in range(epochs_per_batch[b]):
            model.train()
            logits = model(graph, features)
            seen_cols_t = torch.tensor(seen_class_ids, dtype=torch.long, device=device)
            # Slice logits to seen columns and use mapped targets
            loss = loss_fn(logits[train_idx_t][:, seen_cols_t], y_seen_mapped)

            if ewc: loss = loss + ewc.penalty(model)
            grads = torch.autograd.grad(loss, model.parameters(), retain_graph=True, allow_unused=True)
            if si: si.update(grads)

            # Distillation on previously seen outputs
            if (mech in use_distill_for) and (teacher_model is not None) and b>0:
                prev_classes = [c for batch in class_batches[:b] for c in batch]
                prev_ids = [class_to_label[c] for c in prev_classes]
                prev_cols_t = torch.tensor(prev_ids, dtype=torch.long, device=device)
                teacher_soft = soft_targets_probs(teacher_model, graph, features, temperature=distill_T)
                student_probs_prev = torch.softmax(model(graph, features), dim=1)[:, prev_cols_t]
                dloss = distillation_kl(student_probs_prev, teacher_soft[:, prev_cols_t], temperature=distill_T)
                loss = (1-distill_alpha)*loss + distill_alpha*dloss

            opt.zero_grad(); loss.backward(); opt.step()

        if si: si.consolidate()
        if ewc: ewc.compute_fisher(graph, features, labels, train_idx)
        if buffer: buffer.add(idx)

        acc, cm, pr, rc, f1, report, class_names_seen = eval_cumulative_dynamic(
            model, features, labels, b, class_batches, label_encoder, class_to_label, all_labels_np
        )
        batch_accs.append(acc); batch_confusions.append(cm); batch_reports.append(report); batch_classnames.append(class_names_seen)

    return {"config":{"mech":mech,"hidden_feats":hidden_feats,"dropout":dropout,"lr":lrs[0],
                      "weight_decay":weight_decay,"epochs_per_batch":sum(epochs_per_batch),
                      "buffer_size":buffer_size,"ewc_lam":ewc_lam,"si_c":si_c},
            "batch_accs":batch_accs,"batch_reports":batch_reports,
            "batch_confusions":batch_confusions,"batch_classnames":batch_classnames}

# -------------------------
# Search spaces
# -------------------------
MECHANISMS = ["naive","rehearsal","ewc","si","gen_replay"]
HSPACE = {
    "hidden_feats": [64, 128, 256],
    "dropout": [0.0, 0.3, 0.5],
    "lr": [1e-3, 5e-4, 1e-4],
    "weight_decay": [0.0, 1e-5, 1e-4],
    "epochs_per_batch": [10, 15, 20],
    "buffer_size": [2000, 5000, 10000],
    "ewc_lam": [500.0, 1000.0, 2000.0],
    "si_c": [2.0, 5.0, 10.0],
}
def sample_random():
    return {
        "hidden_feats": random.choice(HSPACE["hidden_feats"]),
        "dropout": random.choice(HSPACE["dropout"]),
        "lr": random.choice(HSPACE["lr"]),
        "weight_decay": random.choice(HSPACE["weight_decay"]),
        "epochs_per_batch": random.choice(HSPACE["epochs_per_batch"]),
        "buffer_size": random.choice(HSPACE["buffer_size"]),
        "ewc_lam": random.choice(HSPACE["ewc_lam"]),
        "si_c": random.choice(HSPACE["si_c"]),
    }

# -------------------------
# Objective for dynamic setup
# -------------------------
def objective_dynamic(config, mech):
    # Normalize config: scalar lr -> lrs tuple; scalar epochs -> 6-tuple
    lr = config.get("lr", 5e-4)
    ep = config.get("epochs_per_batch", 10)

    lrs = config.get("lrs", (lr,)*6)
    if isinstance(lrs, (float, int)):
        lrs = (float(lrs),)*6

    if isinstance(ep, (int, float)):
        epochs_per_batch = (int(ep),)*6
    else:
        epochs_per_batch = tuple(ep)

    res = run_experiment_dynamic(
        mech=("rehearsal" if mech=="gen_replay" else mech),
        hidden_feats=config.get("hidden_feats", 128),
        dropout=config.get("dropout", 0.5),
        lrs=lrs,
        weight_decay=config.get("weight_decay", 1e-4),
        epochs_per_batch=epochs_per_batch,
        buffer_size=config.get("buffer_size", 5000),
        ewc_lam=config.get("ewc_lam", 1000.0),
        si_c=config.get("si_c", 5.0),
        use_distill_for=("ewc","si"),
        distill_T=2.0,
        distill_alpha=0.5
    )
    avg_acc = float(np.mean(res["batch_accs"]))
    return avg_acc, res

# -------------------------
# Six search methods (dynamic)
# -------------------------
def grid_search_dynamic(max_trials_per_mech=12):
    trials = []
    for mech in  tqdm(MECHANISMS, desc="Grid search mechanisms"):
        combos = []
        for hf in HSPACE["hidden_feats"]:
            for dr in [0.3, 0.5]:
                for lr in [1e-3, 5e-4]:
                    for wd in [0.0, 1e-4]:
                        for ep in [10, 15]:
                            cfg = {"hidden_feats": hf, "dropout": dr, "lr": lr,
                                   "weight_decay": wd, "epochs_per_batch": (ep,)*6,
                                   "buffer_size": 5000, "ewc_lam": 1000.0, "si_c": 5.0}
                            combos.append(cfg)
        for cfg in combos[:max_trials_per_mech]:
            avg_acc, res = objective_dynamic(cfg, mech)
            trials.append({"method":"grid","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
    return trials

def random_search_dynamic(trials_per_mech=12):
    trials = []
    for mech in tqdm(MECHANISMS, desc="Random search mechanisms"):
        for _ in range(trials_per_mech):
            cfg = sample_random()
            cfg["epochs_per_batch"] = (cfg["epochs_per_batch"],)*6
            avg_acc, res = objective_dynamic(cfg, mech)
            trials.append({"method":"random","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
    return trials

def bayes_opt_dynamic(trials_per_mech=15):
    trials = []
    history = {m: [] for m in tqdm(MECHANISMS, desc="Bayesian search mechanisms")}
    for mech in MECHANISMS:
        for t in range(trials_per_mech):
            if t < 5:
                cfg = sample_random()
            else:
                past = sorted(history[mech], key=lambda x: -x["avg_acc"])
                top = past[:max(1, len(past)//3)]
                if top:
                    base = random.choice(top)["config"]
                    cfg = {
                        "hidden_feats": random.choice([base["hidden_feats"]] + HSPACE["hidden_feats"]),
                        "dropout": random.choice([base["dropout"], 0.3, 0.5]),
                        "lr": random.choice([base["lr"], 5e-4, 1e-4]),
                        "weight_decay": random.choice([base["weight_decay"], 1e-5, 1e-4]),
                        "epochs_per_batch": random.choice([base["epochs_per_batch"][0], 15, 20]),
                        "buffer_size": random.choice([base["buffer_size"], 5000, 10000]),
                        "ewc_lam": random.choice([base["ewc_lam"], 1000.0, 2000.0]),
                        "si_c": random.choice([base["si_c"], 5.0, 10.0]),
                    }
            cfg["epochs_per_batch"] = (cfg["epochs_per_batch"],)*6
            avg_acc, res = objective_dynamic(cfg, mech)
            rec = {"method":"bayes","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res}
            trials.append(rec); history[mech].append(rec)
    return trials

def hyperband_dynamic(R=27, eta=3):
    trials = []
    for mech in tqdm(MECHANISMS, desc="Hyperband search mechanisms"):
        s_max = int(math.log(R, eta))
        for s in range(s_max, -1, -1):
            n = int(math.ceil((s_max+1)/(s+1)) * (eta**s))
            r = int(R * (eta**(-s)))
            configs = [sample_random() for _ in range(n)]
            for cfg in configs:
                cfg["epochs_per_batch"] = (r,)*6
            records = []
            for cfg in configs:
                avg_acc, res = objective_dynamic(cfg, mech)
                records.append({"method":"hyperband","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
            k = max(1, n//eta)
            trials.extend(sorted(records, key=lambda x: -x["avg_acc"])[:k])
    return trials

def pbt_dynamic(generations=2, population_size=6, mutate_prob=0.5):
    trials = []
    for mech in tqdm(MECHANISMS, desc="PBT search mechanisms"):
        population = []
        for _ in range(population_size):
            cfg = sample_random(); cfg["epochs_per_batch"] = (cfg["epochs_per_batch"],)*6
            avg_acc, res = objective_dynamic(cfg, mech)
            population.append({"method":"pbt","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
        trials.extend(population)
        for _ in range(generations):
            population = sorted(population, key=lambda x: -x["avg_acc"])
            elites = population[:population_size//2]
            offspring = []
            for e in elites:
                cfg = e["config"].copy()
                if random.random() < mutate_prob: cfg["lr"] = random.choice(HSPACE["lr"])
                if random.random() < mutate_prob: cfg["hidden_feats"] = random.choice(HSPACE["hidden_feats"])
                if random.random() < mutate_prob: cfg["dropout"] = random.choice(HSPACE["dropout"])
                if random.random() < mutate_prob:
                    ep = random.choice(HSPACE["epochs_per_batch"]); cfg["epochs_per_batch"] = (ep,)*6
                if random.random() < mutate_prob: cfg["weight_decay"] = random.choice(HSPACE["weight_decay"])
                if random.random() < mutate_prob and mech in ["rehearsal","gen_replay"]:
                    cfg["buffer_size"] = random.choice(HSPACE["buffer_size"])
                if random.random() < mutate_prob and mech=="ewc":
                    cfg["ewc_lam"] = random.choice(HSPACE["ewc_lam"])
                if random.random() < mutate_prob and mech=="si":
                    cfg["si_c"] = random.choice(HSPACE["si_c"])
                avg_acc, res = objective_dynamic(cfg, mech)
                offspring.append({"method":"pbt","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
            population = elites + offspring
            trials.extend(offspring)
    return trials

def ga_dynamic(generations=2, population_size=8, crossover_rate=0.7, mutation_rate=0.2):
    trials = []
    for mech in tqdm(MECHANISMS, desc="Evolutionary search mechanisms"):
        pop = []
        for _ in range(population_size):
            cfg = sample_random(); cfg["epochs_per_batch"] = (cfg["epochs_per_batch"],)*6
            avg_acc, res = objective_dynamic(cfg, mech)
            pop.append({"method":"ga","mech":mech,"config":cfg,"avg_acc":avg_acc,"res":res})
        trials.extend(pop)
        for _ in range(generations):
            pop = sorted(pop, key=lambda x: -x["avg_acc"])
            new_pop = pop[:2]
            while len(new_pop) < population_size:
                p1, p2 = random.choice(pop[:len(pop)//2]), random.choice(pop[len(pop)//2:])
                child_cfg = p1["config"].copy()
                if random.random() < crossover_rate:
                    for k in child_cfg:
                        if random.random() < 0.5:
                            child_cfg[k] = p2["config"][k]
                for k in child_cfg:
                    if random.random() < mutation_rate:
                        child_cfg[k] = sample_random()[k]
                if isinstance(child_cfg["epochs_per_batch"], int):
                    child_cfg["epochs_per_batch"] = (child_cfg["epochs_per_batch"],)*6
                avg_acc, res = objective_dynamic(child_cfg, mech)
                new_pop.append({"method":"ga","mech":mech,"config":child_cfg,"avg_acc":avg_acc,"res":res})
            pop = new_pop
            trials.extend(new_pop[2:])
    return trials

# -------------------------
# Run all search methods (dynamic)
# -------------------------
start_time = time.time()
results_trials = []
results_trials += grid_search_dynamic(max_trials_per_mech=12)
results_trials += random_search_dynamic(trials_per_mech=12)
results_trials += bayes_opt_dynamic(trials_per_mech=15)
results_trials += hyperband_dynamic(R=27, eta=3)
results_trials += pbt_dynamic(generations=2, population_size=6)
results_trials += ga_dynamic(generations=2, population_size=8)
elapsed = time.time() - start_time
print(f"\nDynamic search completed in {elapsed/60:.1f} minutes. Total trials: {len(results_trials)}")

# -------------------------
# Consolidated reporting (same schema as static)
# -------------------------
summary_rows = []
for t in results_trials:
    res = t["res"]
    config = res["config"].copy()
    config.update({"method": t["method"], "mech": t["mech"]})
    row = {
        **config,
        "avg_acc": float(np.mean(res["batch_accs"])),
        "batch1_acc": res["batch_accs"][0],
        "batch2_acc": res["batch_accs"][1],
        "batch3_acc": res["batch_accs"][2],
        "batch4_acc": res["batch_accs"][3],
        "batch5_acc": res["batch_accs"][4],
        "batch6_acc": res["batch_accs"][5],
    }
    summary_rows.append(row)

df_summary = pd.DataFrame(summary_rows)
df_summary.to_csv("gcn_dynamic_summary.csv", index=False)
print("Saved gcn_dynamic_summary.csv")

print("\n=== Top-5 configurations per search method (by avg cumulative accuracy) ===")
for method in ["grid","random","bayes","hyperband","pbt","ga"]:
    df_top = df_summary[df_summary["method"]==method].sort_values("avg_acc", ascending=False).head(5)
    print(f"\n[{method.upper()}] Top-5")
    print(tabulate(df_top.round(3), headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))

df_overall_top = df_summary.sort_values("avg_acc", ascending=False).head(10)
print("\n=== Overall Top-10 configurations ===")
print(tabulate(df_overall_top.round(3), headers="keys", tablefmt="grid", showindex=False, floatfmt=".3f"))

print("\n=== Detailed classification reports for best config per mechanism ===")
for mech in MECHANISMS:
    mech_trials = [t for t in results_trials if t["mech"]==mech]
    if not mech_trials: continue
    best = max(mech_trials, key=lambda x: x["avg_acc"])
    res = best["res"]
    cfg = best["config"]
    print(f"\nMechanism: {mech} | Method: {best['method']} | AvgAcc: {np.mean(res['batch_accs']):.3f}")
    print(f"Config: {cfg}")
    for b in range(len(class_batches)):
        classes_seen = ", ".join([c for batch in class_batches[:b+1] for c in batch])
        print(f"\n- Batch {b+1} (seen: {classes_seen}) | Acc: {res['batch_accs'][b]:.3f}")
        print("Confusion matrix:")
        cm_df = pd.DataFrame(res["batch_confusions"][b],
                             index=res["batch_classnames"][b],
                             columns=res["batch_classnames"][b])
        print(cm_df)
        print("\nClassification report:")
        print(res["batch_reports"][b])
```
